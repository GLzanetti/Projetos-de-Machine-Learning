# Aula 1 de Machine Learning
## Regressão Linear Simples do Zero (Sem Bibliotecas)

Este notebook apresenta a implementação de um modelo documento supervisionado de **Regressão Linear Simples** desenvolvido de forma puramente matemática e algorítmica, sem a utilização de bibliotecas de Machine Learning (como Scikit-Learn).

## 📋 Objetivo
Encontrar os melhores parâmetros de inclinação (peso/slope $w$) e intercepto (viés/bias $b$) que minimizam a diferença entre os valores previstos e os valores reais em um pequeno conjunto de dados.

## 🛠️ Detalhes da Implementação
1. **Previsão**: Utiliza a função clássica da reta:  
   $$y = w \cdot x + b$$

2. **Métrica de Avaliação**: **Erro Quadrático Médio (MSE)** para medir a qualidade do ajuste:
   $$MSE = \frac{1}{N} \sum_{i=1}^{N} (y_{previsto} - y_{real})^2$$

3. **Método de Otimização**: **Busca em Grade (Grid Search)** simples por força bruta, testando combinações inteiras de parâmetros em um intervalo pré-definido.

```python
def prever(w, x, b):
  yChapeu = w*x + b
  return yChapeu

def equacaoDoErro(xDados:list, yDados:list, w, b):
  somaErrosQuadrados = 0

  for x, yVerdadeiro in zip(xDados, yDados):
    yChapeu = prever(w, x, b)
    erroQuadrado = (yChapeu - yVerdadeiro)**2
    somaErrosQuadrados += erroQuadrado
    quantidadeDados = len(xDados)

  mse = somaErrosQuadrados / quantidadeDados

  return mse

```

```python
melhorPeso = None
melhorBias = None
menorErro = float('inf') #Numero que pode chegar ao infinito

xDados = [1, 2, 3]
yDados = [3, 5, 7]

for w in range(-11, 10):
  for b in range(-11, 10):
    erro = equacaoDoErro(xDados, yDados, w, b)
    if erro < menorErro:
      menorErro = erro
      melhorPeso = w
      melhorBias = b
      print("Novo melhor modelo: ")
      print(f"w = {w} b = {b} MSE = {erro:.2f}")
```
