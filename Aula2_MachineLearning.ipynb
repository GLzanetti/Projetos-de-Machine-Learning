# Aula 2 de Machine Learning
## Regressão Linear Simples do Zero (Sem Bibliotecas)

Este notebook apresenta a implementação de um modelo supervisionado de **Regressão Linear Simples** desenvolvido de forma puramente matemática e algorítmica, sem a utilização de bibliotecas de Machine Learning (como Scikit-Learn).

## 📋 Objetivo
Encontrar os melhores parâmetros de inclinação (peso $w$) e intercepto (viés $b$) que minimizam a diferença entre os valores previstos e os valores reais de forma automatizada sobre um conjunto de dados.

## 🛠️ Detalhes da Implementação
1. **Previsão**: Utiliza a função clássica da reta:
   $$y = w \cdot x + b$$

2. **Métrica de Avaliação**: **Erro Quadrático Médio (MSE)** para medir a qualidade do ajuste a cada época:
   $$MSE = \frac{1}{n} \sum_{i=1}^{n} (y_{previsto} - y_{real})^2$$

3. **Método de Otimização**: **Gradiente Descendente**, ajustando os parâmetros $w$ e $b$ de forma iterativa ao calcular as derivadas parciais do erro (gradientes) a cada época de treino.

```python
def equacaoRL(w,x,b):
  yC = w*x + b
  return yC

def calcMSE(xdados:list,ydados:list,w,b):
  somaErrosQuadrados = 0

  for x,y in zip(xdados,ydados):
    yC = equacaoRL(w,x,b)
    erro = (yC - y)
    erroQuadrado = erro**2
    somaErrosQuadrados += erroQuadrado

  mse = somaErrosQuadrados / len(xdados)
  return mse

def gradiente(xdados:list,ydados:list,w,b):
  gradienteW  = 0
  gradienteB  = 0

  quantidadeDados = len(xdados)

  for x,y in zip(xdados,ydados):
    yC = equacaoRL(w,x,b)
    erro = (yC - y)

    #Calculo de Gradientes de MSE em relação a w
    gradienteW += 2*x*erro

    #Calculo de Gradiente de MSE em relação a b
    gradienteB += 2*erro

    #Como a MSE 1/n
  gradienteW = gradienteW / quantidadeDados
  gradienteB = gradienteB / quantidadeDados

  return gradienteW, gradienteB

def treinar(xdados:list,ydados:list,learningRate=0.01,epocas=1000):
  w = 0
  b = 0

  for epoca in range(epocas):
    gradienteW, gradienteB = gradiente(xdados,ydados,w,b)

    w = w - learningRate*gradienteW
    b = b - learningRate*gradienteB

    mse = calcMSE(xdados,ydados,w,b)

    if epoca % 100 == 0: #A cada 100 épocas o modelo mostra o quanto melhorou
      print(f'Época {epoca:4d}', f'w = {w:.6f}', f'b = {b:.6f}', f'MSE = {mse:.8f}')

    if mse<0.000001:
      print(f'O modelo convergiu após {epoca} épocas')
      break

  return w,b
```

```python
x = [3,6,9]
y = [6,9,12]

wFinal, bFinal = treinar(x,y,epocas=15000,learningRate=0.001)

print("Resultado final: ")
print(f"w = {wFinal:.6f}")
print(f"b = {bFinal:.6f}")

print("Modelo encontrado: ")
print(f"y = {wFinal:.4f}x + {bFinal:.4f}")
```
